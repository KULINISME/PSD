# Land Cover / LULC
Land Cover atau tutupan lahan merujuk pada objek fisik yang menutupi permukaan bumi, seperti sawah, hutan,
bangunan, perairan, dan vegetasi pesisir. Sementara itu, Land Use/Land Cover (LULC) merupakan kombinasi
antara kondisi tutupan lahan dan pemanfaatan lahan yang dilakukan manusia terhadap wilayah tersebut. Dalam
konteks penginderaan jauh, LULC digunakan untuk mengklasifikasikan permukaan bumi secara spasial 
berdasarkan karakteristik spektral citra satelit.

Di Jawa, pola tutupan lahan sangat dipengaruhi oleh aktivitas manusia, kondisi topografi, serta keberadaan
wilayah pesisir dan pertanian. Sawah dan lahan pertanian umumnya tersebar di dataran rendah dan lembah,
sedangkan kawasan hutan dan vegetasi alami cenderung berada di wilayah pegunungan atau daerah konservasi.
Sementara itu, area bangunan dan permukiman berkembang pesat di sekitar kota, jalur transportasi, dan pusat
aktivitas ekonomi. Perairan seperti laut, danau, dan mangrove memainkan peran penting dalam menjaga
ekosistem pesisir serta keseimbangan lingkungan.

Pemanfaatan data LULC sangat penting untuk mendukung perencanaan tata ruang, pengawasan lingkungan,
evaluasi perubahan lahan, serta pengambilan keputusan di bidang pertanian, kehutanan, dan pengelolaan
wilayah pesisir. Dengan pendekatan klasifikasi berbasis citra satelit, perubahan tutupan lahan dapat
dipantau secara periodik, sehingga dapat digunakan untuk analisis spasial yang lebih akurat dan
berkelanjutan.
